# ***=====DATA VISUALIZATION=====***

```python
import matplotlib.pyplot as plt
```

# **Horizontal Bar Chart**

```python
top_10_happiness = (
    confirmed_df.groupby('Country')['Happiness_Score']
    .mean()
    .sort_values(ascending=False)
    .head(10))

top_10_happiness = top_10_happiness.sort_values(ascending=True)

plt.figure(figsize=(10, 6))

colors = [
    '#A8DADC',
    '#FFD6A5',
    '#BDE0FE',
    '#CDEAC0',
    '#D8B4FE',
    '#FFAFCC',
    '#B8F2E6',
    '#FFF3B0',
    '#CDB4DB',
    '#F1C0E8'
]

bars = plt.barh(
    top_10_happiness.index,
    top_10_happiness.values,
    color=colors)

plt.title("Top 10 Happiest Countries", fontsize=16, fontweight='bold')
plt.xlabel("Average Happiness Score", fontsize=12)
plt.ylabel("Country", fontsize=12)

for bar in bars:
    plt.text(
        bar.get_width() + 0.02,
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():.2f}",
        va='center',
        fontsize=10)

plt.xlim(0, 8)
plt.tight_layout()
plt.show()
```

# **Bar Chart (Happiness Score)**

1. The bar chart shows the top 10 happiest countries based on their average Happiness Score.

2. Finland has the highest average Happiness Score among the selected countries.

3. Other countries also show high Happiness Scores with relatively small differences.

4. The chart provides a clear comparison of happiness levels across different countries.

5. This visualization helps to easily identify and compare the happiest countries.

# **Pie Chart confirmed-cases**

```python
top5 = (
    confirmed_df.groupby('Country')['Happiness_Score']
    .mean()
    .sort_values(ascending=False)
    .head(5))

plt.figure(figsize=(8, 8))

explode = (0.1, 0, 0, 0, 0)

plt.pie(
    top5.values,
    labels=top5.index,
    autopct='%1.1f%%',
    explode=explode,
    startangle=90)

plt.title(
    "Top 5 Happiest Countries by Average Happiness Score",
    fontsize=14,
    fontweight='bold')

plt.show()
```

# **Pie Chart (Confirmed Cases)**

1.The pie chart shows the percentage share of the average happiness score among the top 5 happiest countries.

2.Finland contributes the largest proportion of the total happiness score among the selected top 5 countries.

3.Denmark, Iceland, Norway, and Switzerland account for the remaining shares.

4.The exploded slice highlights Finland, which has the highest average happiness score.

5.This chart helps visualize how the happiness scores are distributed among the top 5 happiest countries.

# **Scatter plot = Happiness score**

```python
import matplotlib.pyplot as plt
country_happiness = (
    confirmed_df.groupby('Country')['Happiness_Score']
    .mean())

top10 = country_happiness.sort_values(ascending=False).head(10)
plt.figure(figsize=(12, 6))

plt.plot(
    top10.index,
    top10.values,
    marker='o',
    linewidth=2)

plt.title("Top 10 Countries by Happiness Score")
plt.xlabel("Country")
plt.ylabel("Average Happiness Score")
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# **Scatter Plot (Happiness Score)**
1.The scatter plot displays the average Happiness Score for the top 10 happiest countries.

2.Each point represents the average Happiness Score of one country.

3.Finland appears at the highest position with the highest average Happiness Score among the selected countries.

4.The other countries are positioned close to Finland, showing comparatively high Happiness Scores.

5.The scatter plot effectively highlights the differences in Happiness Scores among the top 10 happiest countries.

# **COVID-19 Dashboard (Confirmed Cases)**

```python
import matplotlib.pyplot as plt
country_happiness = (
    confirmed_df.groupby('Country')['Happiness_Score']
    .mean())

top10 = country_happiness.sort_values(ascending=False).head(10)

top5 = country_happiness.sort_values(ascending=False).head(5)

plt.figure(figsize=(18, 10))

plt.subplot(2, 3, 1)

colors = [
    '#A8DADC', '#FFD6A5', '#BDE0FE', '#CDEAC0', '#D8B4FE',
    '#FFAFCC', '#B8F2E6', '#FFF3B0', '#CDB4DB', '#F1C0E8'
]

bars = plt.bar(
    top10.index,
    top10.values,
    color=colors)

plt.title("Top 10 Happiest Countries")
plt.ylabel("Average Happiness Score")
plt.xticks(rotation=45, ha='right')

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.03,
        f"{bar.get_height():.2f}",
        ha='center',
        fontsize=9)

plt.subplot(2, 3, 2)

plt.scatter(
    top10.index,
    top10.values,
    s=120,
    marker='o')

plt.title("Happiness Score Scatter Plot")
plt.ylabel("Average Happiness Score")
plt.xticks(rotation=45, ha='right')
plt.grid(True, linestyle='--', alpha=0.5)

plt.subplot(2, 3, 3)

plt.plot(
    top10.index,
    top10.values,
    marker='o')

plt.title("Happiness Score Line Chart")
plt.ylabel("Average Happiness Score")
plt.xticks(rotation=45, ha='right')
plt.grid(True, linestyle='--', alpha=0.5)
plt.subplot(2, 3, 4)

top10_horizontal = top10.sort_values(ascending=True)

bars = plt.barh(
    top10_horizontal.index,
    top10_horizontal.values,
    color=colors)

plt.title("Top 10 Happiest Countries")
plt.xlabel("Average Happiness Score")

for bar in bars:
    plt.text(
        bar.get_width() + 0.02,
        bar.get_y() + bar.get_height() / 2,
        f"{bar.get_width():.2f}",
        va='center',
        fontsize=9)

plt.subplot(2, 3, 5)

explode = (0.1, 0, 0, 0, 0)

plt.pie(
    top5.values,
    labels=top5.index,
    autopct='%1.1f%%',
    explode=explode,
    shadow=True,
    startangle=90
)

plt.title("Top 5 Happiest Countries")

plt.suptitle(
    "Global Happiness Report Analysis Dashboard",
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
```

# **Dashboard (Combined Visualization)**

1.The dashboard combines multiple visualizations of Happiness Scores in a single view for better analysis.

2.It allows quick comparison of the top happiest countries using different chart types.

3.The dashboard highlights the countries with the highest average Happiness Scores.

4.Different visualizations provide insights into rankings, comparisons, proportions, and score differences among countries.

5.Overall, the dashboard provides a comprehensive overview of the Global Happiness dataset, making the analysis easier to understand.
